```python
DATA_DIR = Path("/content/drive/MyDrive/Language Detection")
PARQUET_PATH = DATA_DIR / "sessions_lang_transcript_2026-08-13_2026-08-14.parquet"

TARGET_LANGUAGES = {
    "en": ("English"),
    "es": ("Spanish"),
    "de": ("German"),
    "fr": ("French"),
    "ru": ("Russian"),
    "pt": ("Portuguese"),
}

SEGMENTS_PER_LANGUAGE = 25

if not PARQUET_PATH.exists():
    raise FileNotFoundError(PARQUET_PATH)

con = duckdb.connect()

con.execute(
    f"""
    CREATE OR REPLACE VIEW sessions AS
    SELECT *
    FROM read_parquet('{PARQUET_PATH.as_posix()}')
    """
)

session_stats = con.execute(
    """
    SELECT
        COUNT(*) AS rows,
        COUNT(DISTINCT gamesession_id) AS unique_sessions
    FROM sessions
    """
).df()

display(session_stats)
```

```python
language_metadata = pd.DataFrame(
    [
        {
            "lang_detected": code,
            "language_name": values[0],
        }
        for code, values in TARGET_LANGUAGES.items()
    ]
)

total_sessions = int(session_stats.loc[0, "unique_sessions"])

top_6_languages = con.execute(
    """
    SELECT
        lang_detected,
        COUNT(DISTINCT gamesession_id) AS session_count
    FROM sessions
    GROUP BY lang_detected
    ORDER BY session_count DESC
    LIMIT 6
    """
).df()

top_6_languages = (
    top_6_languages
    .merge(language_metadata, on="lang_detected", how="left")
    .assign(
        percentage=lambda x: (
            x["session_count"] / total_sessions * 100
        ).round(2)
    )
    [
        [
            "lang_detected",
            "language_name",
            "session_count",
            "percentage",
        ]
    ]
)

top_6_languages["percentage"] = (
    top_6_languages["percentage"].astype(str) + "%"
)

display(top_6_languages)
```

```python
original_head = con.execute(
    """
    SELECT *
    FROM sessions
    LIMIT 5
    """
).df()

display(original_head)
```

```python
target_codes = tuple(TARGET_LANGUAGES)

language_counts = con.execute(
    f"""
    SELECT
        lang_detected,
        COUNT(DISTINCT gamesession_id) AS session_count
    FROM sessions
    WHERE lang_detected IN {target_codes}
    GROUP BY lang_detected
    ORDER BY session_count DESC
    """
).df()

language_counts = (
    language_metadata
    .merge(language_counts, on="lang_detected", how="left")
    .fillna({"session_count": 0})
)

language_counts["session_count"] = language_counts["session_count"].astype(int)

language_counts["percentage"] = (
    language_counts["session_count"] / total_sessions * 100
).round(2)

language_counts = (
    language_counts
    .sort_values("session_count", ascending=False)
    .reset_index(drop=True)
    [
        [
            "lang_detected",
            "language_name",
            "session_count",
            "percentage",
        ]
    ]
)

language_counts["percentage"] = (
    language_counts["percentage"].astype(str) + "%"
)

print("Total sessions:", f"{total_sessions:,}")
display(language_counts)
```

```python
target_codes = tuple(TARGET_LANGUAGES)

latest_25_segments = con.execute(
    f"""
    WITH target_sessions AS (
        SELECT
            gamesession_id,
            user_id,
            game_name,
            model_type,
            TRY_CAST(created_at AS TIMESTAMP) AS session_created_at,
            lang_detected,
            lang_probability,
            transcript_segments
        FROM sessions
        WHERE lang_detected IN {target_codes}
    ),
    exploded AS (
        SELECT
            s.gamesession_id,
            s.user_id,
            s.game_name,
            s.model_type,
            s.session_created_at,
            s.lang_detected,
            s.lang_probability,
            segment.text AS segment_text,
            TRY_CAST(segment.timestamp[1] AS DOUBLE) AS segment_start_seconds,
            TRY_CAST(segment.timestamp[2] AS DOUBLE) AS segment_end_seconds
        FROM target_sessions AS s
        CROSS JOIN UNNEST(s.transcript_segments) AS u(segment)
    ),
    ranked AS (
        SELECT
            *,
            session_created_at
                + COALESCE(segment_start_seconds, 0) * INTERVAL 1 SECOND
                AS segment_created_at,
            ROW_NUMBER() OVER (
                PARTITION BY lang_detected
                ORDER BY
                    session_created_at
                        + COALESCE(segment_start_seconds, 0) * INTERVAL 1 SECOND DESC,
                    gamesession_id DESC
            ) AS language_rank
        FROM exploded
        WHERE
            segment_text IS NOT NULL
            AND TRIM(segment_text) <> ''
            AND session_created_at IS NOT NULL
    )
    SELECT
        lang_detected,
        gamesession_id,
        user_id,
        game_name,
        model_type,
        session_created_at,
        segment_created_at,
        segment_start_seconds,
        segment_end_seconds,
        lang_probability,
        segment_text
    FROM ranked
    WHERE language_rank <= {SEGMENTS_PER_LANGUAGE}
    ORDER BY lang_detected, segment_created_at DESC
    """
).df()

latest_25_segments.insert(
    1,
    "language_name",
    latest_25_segments["lang_detected"].map(
        {code: values[0] for code, values in TARGET_LANGUAGES.items()}
    ),
)

display(latest_25_segments)
```

```python
segment_counts = (
    latest_25_segments
    .groupby(["lang_detected", "language_name"])
    .size()
    .reset_index(name="segment_count")
    .sort_values("lang_detected")
    .reset_index(drop=True)
)

order_check = (
    latest_25_segments
    .groupby("lang_detected")["segment_created_at"]
    .apply(lambda values: values.is_monotonic_decreasing)
    .reset_index(name="newest_first")
)

validation = segment_counts.merge(order_check, on="lang_detected")

if len(latest_25_segments) != len(TARGET_LANGUAGES) * SEGMENTS_PER_LANGUAGE:
    raise ValueError("Unexpected segment count")

if not validation["newest_first"].all():
    raise ValueError("Segment ordering validation failed")

display(validation)
print("Total segments:", len(latest_25_segments))
```

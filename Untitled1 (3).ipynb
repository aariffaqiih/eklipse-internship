```python
import pyarrow.parquet as pq
import pandas as pd

PARQUET_PATH = "/content/drive/MyDrive/Language Detection/sessions_lang_transcript_2026-08-13_2026-08-14.parquet"

parquet_file = pq.ParquetFile(PARQUET_PATH)

COLUMNS = [
    "gamesession_id",
    "user_id",
    "game_name",
    "model_type",
    "created_at",
    "lang_detected",
    "lang_probability",
]

chunks = []

for row_group_index in range(parquet_file.num_row_groups):
    chunk = parquet_file.read_row_group(
        row_group_index,
        columns=COLUMNS,
    ).to_pandas()

    chunks.append(chunk)

    print(
        f"Loaded row group "
        f"{row_group_index + 1}/{parquet_file.num_row_groups} "
        f"({len(chunk):,} rows)"
    )

df_sessions = pd.concat(
    chunks,
    ignore_index=True,
)

print()
print("Parquet metadata rows :", f"{parquet_file.metadata.num_rows:,}")
print("Loaded dataframe rows :", f"{len(df_sessions):,}")
print("Unique session IDs    :", f"{df_sessions['gamesession_id'].nunique():,}")

display(df_sessions.head(20))
```

```python
assert len(df_sessions) == parquet_file.metadata.num_rows

print("OK: semua rows dari parquet berhasil terbaca.")
```

```python
total_sessions = df_sessions["gamesession_id"].nunique()

top_6_languages = (
    df_sessions
    .groupby("lang_detected")["gamesession_id"]
    .nunique()
    .sort_values(ascending=False)
    .head(6)
    .reset_index(name="session_count")
)

top_6_languages["percentage"] = (
    top_6_languages["session_count"]
    / total_sessions
    * 100
).round(2)

display(top_6_languages)
```

```python
target_languages = ["en", "de", "ru", "fr", "es", "pt"]

total_sessions = df_sessions["gamesession_id"].nunique()

language_counts = (
    df_sessions[
        df_sessions["lang_detected"].isin(target_languages)
    ]
    .groupby("lang_detected")["gamesession_id"]
    .nunique()
    .reindex(target_languages, fill_value=0)
    .reset_index(name="session_count")
)

language_counts["percentage"] = (
    language_counts["session_count"]
    / total_sessions
    * 100
).round(2)

language_counts["percentage"] = (
    language_counts["percentage"]
    .astype(str)
    + "%"
)

print("Total sessions:", f"{total_sessions:,}")

display(language_counts)
```

```python
language_metadata = {
    "en": {"language_name": "English", "country": "United States / United Kingdom"},
    "de": {"language_name": "German", "country": "Germany"},
    "ru": {"language_name": "Russian", "country": "Russia"},
    "fr": {"language_name": "French", "country": "France"},
    "es": {"language_name": "Spanish", "country": "Spain"},
    "pt": {"language_name": "Portuguese", "country": "Portugal / Brazil"},
    "id": {"language_name": "Indonesian", "country": "Indonesia"},
    "it": {"language_name": "Italian", "country": "Italy"},
    "nl": {"language_name": "Dutch", "country": "Netherlands"},
    "pl": {"language_name": "Polish", "country": "Poland"},
    "uk": {"language_name": "Ukrainian", "country": "Ukraine"},
    "tr": {"language_name": "Turkish", "country": "Türkiye"},
    "ja": {"language_name": "Japanese", "country": "Japan"},
    "ko": {"language_name": "Korean", "country": "South Korea"},
    "zh": {"language_name": "Chinese", "country": "China"},
}

total_sessions = df_sessions["gamesession_id"].nunique()

top_6_languages = (
    df_sessions
    .groupby("lang_detected")["gamesession_id"]
    .nunique()
    .reset_index(name="session_count")
    .sort_values("session_count", ascending=False)
    .head(6)
    .reset_index(drop=True)
)

top_6_languages["language_name"] = top_6_languages["lang_detected"].map(
    lambda x: language_metadata.get(x, {}).get("language_name", "Unknown")
)

top_6_languages["country"] = top_6_languages["lang_detected"].map(
    lambda x: language_metadata.get(x, {}).get("country", "Unknown")
)

top_6_languages["percentage"] = (
    top_6_languages["session_count"]
    / total_sessions
    * 100
).round(2)

top_6_languages["percentage"] = (
    top_6_languages["percentage"].astype(str) + "%"
)

top_6_languages = top_6_languages[
    [
        "lang_detected",
        "language_name",
        "country",
        "session_count",
        "percentage",
    ]
]

display(top_6_languages)
```

```python
target_languages = ["en", "de", "ru", "fr", "es", "pt"]

total_sessions = df_sessions["gamesession_id"].nunique()

language_counts = (
    df_sessions[
        df_sessions["lang_detected"].isin(target_languages)
    ]
    .groupby("lang_detected")["gamesession_id"]
    .nunique()
    .reset_index(name="session_count")
)

language_counts = (
    language_counts
    .set_index("lang_detected")
    .reindex(target_languages, fill_value=0)
    .reset_index()
)

language_counts["language_name"] = language_counts["lang_detected"].map(
    lambda x: language_metadata.get(x, {}).get("language_name", "Unknown")
)

language_counts["country"] = language_counts["lang_detected"].map(
    lambda x: language_metadata.get(x, {}).get("country", "Unknown")
)

language_counts["percentage"] = (
    language_counts["session_count"]
    / total_sessions
    * 100
).round(2)

language_counts = (
    language_counts
    .sort_values("session_count", ascending=False)
    .reset_index(drop=True)
)

language_counts["percentage"] = (
    language_counts["percentage"].astype(str) + "%"
)

language_counts = language_counts[
    [
        "lang_detected",
        "language_name",
        "country",
        "session_count",
        "percentage",
    ]
]

print("Total sessions:", f"{total_sessions:,}")

display(language_counts)
```

```python
%pip -q install -U duckdb

from pathlib import Path
import os

import duckdb
from google.colab import drive

if not Path("/content/drive/MyDrive").exists():
    drive.mount("/content/drive")

PARQUET_PATH = Path(
    "/content/drive/MyDrive/Language Detection/"
    "sessions_lang_transcript_2026-08-23_2026-08-24.parquet"
)

if not PARQUET_PATH.is_file():
    raise FileNotFoundError(PARQUET_PATH)

THREADS = min(os.cpu_count() or 4, 8)

con = duckdb.connect()

con.execute(f"SET threads = {THREADS}")
con.execute("SET preserve_insertion_order = false")
con.execute("SET enable_progress_bar = true")

parquet = PARQUET_PATH.as_posix()

row_count = con.execute(
    "SELECT COUNT(*) FROM read_parquet(?)",
    [parquet],
).fetchone()[0]

schema = con.execute(
    "DESCRIBE SELECT * FROM read_parquet(?)",
    [parquet],
).df()

print(f"File    : {PARQUET_PATH.name}")
print(f"Size    : {PARQUET_PATH.stat().st_size / 1024**2:,.2f} MB")
print(f"Rows    : {row_count:,}")
print(f"Threads : {THREADS}")

display(schema)
```

```python
head = con.execute(
    """
    SELECT
        gamesession_id,
        user_id,
        game_name,
        url,
        model_type,
        created_at,
        lang_detected,
        lang_probability,
        len(transcript_segments) AS segment_count
    FROM read_parquet(?)
    LIMIT 10
    """,
    [parquet],
).df()

display(head)
```

```python
statistics = con.execute(
    """
    SELECT
        COUNT(*) AS total_sessions,

        COUNT(DISTINCT gamesession_id) AS unique_sessions,
        COUNT(DISTINCT user_id) AS unique_users,

        COUNT(DISTINCT game_name) AS unique_games,
        COUNT(DISTINCT model_type) AS unique_model_types,
        COUNT(DISTINCT lang_detected) AS unique_languages,

        ROUND(AVG(lang_probability), 4) AS avg_language_probability,
        ROUND(MIN(lang_probability), 4) AS min_language_probability,
        ROUND(MAX(lang_probability), 4) AS max_language_probability,

        SUM(len(transcript_segments)) AS total_segments,
        MIN(len(transcript_segments)) AS min_segments,
        ROUND(AVG(len(transcript_segments)), 2) AS avg_segments,
        MEDIAN(len(transcript_segments)) AS median_segments,
        MAX(len(transcript_segments)) AS max_segments

    FROM read_parquet(?)
    """,
    [parquet],
).df()

display(statistics)
```

```python
null_summary = con.execute(
    """
    SELECT
        COUNT(*) AS total_rows,

        COUNT(*) FILTER (
            WHERE gamesession_id IS NULL
        ) AS gamesession_id_null,

        COUNT(*) FILTER (
            WHERE user_id IS NULL
        ) AS user_id_null,

        COUNT(*) FILTER (
            WHERE game_name IS NULL
        ) AS game_name_null,

        COUNT(*) FILTER (
            WHERE url IS NULL
        ) AS url_null,

        COUNT(*) FILTER (
            WHERE model_type IS NULL
        ) AS model_type_null,

        COUNT(*) FILTER (
            WHERE created_at IS NULL
        ) AS created_at_null,

        COUNT(*) FILTER (
            WHERE lang_detected IS NULL
        ) AS lang_detected_null,

        COUNT(*) FILTER (
            WHERE lang_probability IS NULL
        ) AS lang_probability_null,

        COUNT(*) FILTER (
            WHERE transcript_segments IS NULL
        ) AS transcript_segments_null

    FROM read_parquet(?)
    """,
    [parquet],
).df()

display(null_summary)
```

```python
segment_statistics = con.execute(
    """
    SELECT
        COUNT(*) AS total_sessions,

        COUNT(*) FILTER (
            WHERE transcript_segments IS NULL
        ) AS null_segment_sessions,

        COUNT(*) FILTER (
            WHERE len(transcript_segments) = 0
        ) AS zero_segment_sessions,

        COUNT(*) FILTER (
            WHERE len(transcript_segments) = 1
        ) AS one_segment_sessions,

        COUNT(*) FILTER (
            WHERE len(transcript_segments) >= 100
        ) AS sessions_ge_100_segments,

        SUM(len(transcript_segments)) AS total_segments,

        MIN(len(transcript_segments)) AS min_segments,

        ROUND(
            AVG(len(transcript_segments)),
            2
        ) AS avg_segments,

        MEDIAN(
            len(transcript_segments)
        ) AS median_segments,

        QUANTILE_CONT(
            len(transcript_segments),
            0.25
        ) AS q25_segments,

        QUANTILE_CONT(
            len(transcript_segments),
            0.75
        ) AS q75_segments,

        MAX(len(transcript_segments)) AS max_segments

    FROM read_parquet(?)
    """,
    [parquet],
).df()

display(segment_statistics)
```

```python
smallest_sessions = con.execute(
    """
    SELECT
        gamesession_id,
        user_id,
        game_name,
        url,
        model_type,
        created_at,
        lang_detected,
        lang_probability,
        len(transcript_segments) AS segment_count
    FROM read_parquet(?)
    ORDER BY
        segment_count ASC NULLS FIRST,
        created_at DESC
    LIMIT 20
    """,
    [parquet],
).df()

display(smallest_sessions)
```

```python
largest_sessions = con.execute(
    """
    SELECT
        gamesession_id,
        user_id,
        game_name,
        url,
        model_type,
        created_at,
        lang_detected,
        lang_probability,
        len(transcript_segments) AS segment_count
    FROM read_parquet(?)
    ORDER BY
        segment_count DESC NULLS LAST,
        created_at DESC
    LIMIT 20
    """,
    [parquet],
).df()

display(largest_sessions)
```

```python
duplicate_sessions = con.execute(
    """
    SELECT
        gamesession_id,
        COUNT(*) AS occurrences
    FROM read_parquet(?)
    GROUP BY gamesession_id
    HAVING COUNT(*) > 1
    ORDER BY occurrences DESC, gamesession_id
    """,
    [parquet],
).df()

print(f"Duplicate gamesession IDs: {len(duplicate_sessions):,}")

display(duplicate_sessions)
```

```python
sessions_100_500 = con.execute(
    """
    WITH source AS (
        SELECT
            gamesession_id,
            transcript_segments
        FROM read_parquet(?)
        WHERE
            transcript_segments IS NOT NULL
            AND len(transcript_segments) BETWEEN 100 AND 500
    ),
    exploded AS (
        SELECT
            gamesession_id,
            UNNEST(transcript_segments) AS segment
        FROM source
    ),
    per_session AS (
        SELECT
            gamesession_id,
            COUNT(*) FILTER (
                WHERE
                    segment.text IS NOT NULL
                    AND TRIM(segment.text) <> ''
                    AND len(
                        regexp_extract_all(
                            TRIM(segment.text),
                            '[\\p{L}\\p{N}]+'
                        )
                    ) >= 4
            ) AS filtered_segment_count
        FROM exploded
        GROUP BY gamesession_id
    )
    SELECT
        COUNT(*) AS session_count
    FROM per_session
    WHERE filtered_segment_count BETWEEN 100 AND 500
    """,
    [parquet],
).df()

display(sessions_100_500)
```

```python
valid_segment_total = con.execute(
    """
    WITH source AS (
        SELECT
            gamesession_id,
            transcript_segments
        FROM read_parquet(?)
        WHERE
            transcript_segments IS NOT NULL
            AND len(transcript_segments) BETWEEN 100 AND 500
    ),
    exploded AS (
        SELECT
            gamesession_id,
            UNNEST(transcript_segments) AS segment
        FROM source
    ),
    per_session AS (
        SELECT
            gamesession_id,
            COUNT(*) FILTER (
                WHERE
                    segment.text IS NOT NULL
                    AND TRIM(segment.text) <> ''
                    AND len(
                        regexp_extract_all(
                            TRIM(segment.text),
                            '[\\p{L}\\p{N}]+'
                        )
                    ) >= 4
            ) AS valid_segment_count
        FROM exploded
        GROUP BY gamesession_id
    )
    SELECT
        COUNT(*) AS valid_sessions,
        SUM(valid_segment_count) AS total_valid_segments
    FROM per_session
    WHERE valid_segment_count BETWEEN 100 AND 500
    """,
    [parquet],
).df()

display(valid_segment_total)
```

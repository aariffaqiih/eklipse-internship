```python
import pyarrow.parquet as pq
import pandas as pd

PARQUET_PATH = "/content/drive/MyDrive/Language Detection/sessions_lang_transcript_2026-08-13_2026-08-14.parquet"

parquet_file = pq.ParquetFile(PARQUET_PATH)

COLUMNS = [
    "gamesession_id",
    "user_id",
    "game_name",
    "model_type",
    "created_at",
    "lang_detected",
    "lang_probability",
]

chunks = []

for row_group_index in range(parquet_file.num_row_groups):
    chunk = parquet_file.read_row_group(
        row_group_index,
        columns=COLUMNS,
    ).to_pandas()

    chunks.append(chunk)

    print(
        f"Loaded row group "
        f"{row_group_index + 1}/{parquet_file.num_row_groups} "
        f"({len(chunk):,} rows)"
    )

df_sessions = pd.concat(
    chunks,
    ignore_index=True,
)

print()
print("Parquet metadata rows :", f"{parquet_file.metadata.num_rows:,}")
print("Loaded dataframe rows :", f"{len(df_sessions):,}")
print("Unique session IDs    :", f"{df_sessions['gamesession_id'].nunique():,}")

display(df_sessions.head(20))
```

```python
import pandas as pd
import pyarrow.parquet as pq

TARGET_LANGUAGES = {
    "en": "English",
    "es": "Spanish",
    "de": "German",
    "fr": "French",
    "ru": "Russian",
    "pt": "Portuguese",
}

SEGMENTS_PER_LANGUAGE = 25
CANDIDATE_SESSIONS_PER_LANGUAGE = 25

df_sessions["created_at"] = pd.to_datetime(
    df_sessions["created_at"],
    errors="coerce",
)

candidate_sessions = (
    df_sessions[
        df_sessions["lang_detected"].isin(TARGET_LANGUAGES)
    ]
    .sort_values("created_at", ascending=False)
    .groupby("lang_detected", group_keys=False)
    .head(CANDIDATE_SESSIONS_PER_LANGUAGE)
)

candidate_session_ids = (
    candidate_sessions["gamesession_id"]
    .drop_duplicates()
    .tolist()
)

candidate_session_ids_set = set(candidate_session_ids)

transcript_table = parquet_file.read(
    columns=[
        "gamesession_id",
        "transcript_segments",
    ]
)

transcript_df = transcript_table.to_pandas()

transcript_df = transcript_df[
    transcript_df["gamesession_id"].isin(candidate_session_ids_set)
]

candidate_sessions = candidate_sessions.merge(
    transcript_df,
    on="gamesession_id",
    how="left",
)

segment_rows = []

for _, row in candidate_sessions.iterrows():

    segments = row["transcript_segments"]

    if segments is None:
        continue

    for segment_index, segment in enumerate(segments, start=1):

        text = str(segment.get("text", "")).strip()

        if not text:
            continue

        timestamp = segment.get("timestamp")

        start_seconds = 0.0
        end_seconds = 0.0

        if timestamp is not None and len(timestamp) >= 2:
            start_seconds = float(timestamp[0])
            end_seconds = float(timestamp[1])

        segment_started_at = (
            row["created_at"]
            + pd.to_timedelta(start_seconds, unit="s")
        )

        segment_ended_at = (
            row["created_at"]
            + pd.to_timedelta(end_seconds, unit="s")
        )

        segment_rows.append(
            {
                "language": row["lang_detected"],
                "language_name": TARGET_LANGUAGES[
                    row["lang_detected"]
                ],
                "gamesession_id": row["gamesession_id"],
                "user_id": row["user_id"],
                "game_name": row["game_name"],
                "session_created_at": row["created_at"],
                "segment_index": segment_index,
                "segment_started_at": segment_started_at,
                "segment_ended_at": segment_ended_at,
                "text": text,
            }
        )

segments_df = pd.DataFrame(segment_rows)

latest_25_segments = (
    segments_df
    .sort_values(
        "segment_started_at",
        ascending=False,
    )
    .groupby(
        "language",
        group_keys=False,
    )
    .head(SEGMENTS_PER_LANGUAGE)
    .sort_values(
        ["language", "segment_started_at"],
        ascending=[True, False],
    )
    .reset_index(drop=True)
)

display(latest_25_segments)
```

```python
import duckdb
import pandas as pd

PARQUET_PATH = "/content/drive/MyDrive/Language Detection/sessions_lang_transcript_2026-08-13_2026-08-14.parquet"

query = f"""
WITH exploded AS (
    SELECT
        p.gamesession_id,
        p.user_id,
        p.game_name,
        p.model_type,
        CAST(p.created_at AS TIMESTAMP) AS session_created_at,
        p.lang_detected,
        p.lang_probability,
        s.segment.text AS segment_text,
        s.segment.timestamp[1] AS segment_start_seconds,
        s.segment.timestamp[2] AS segment_end_seconds
    FROM read_parquet('{PARQUET_PATH}') AS p
    CROSS JOIN UNNEST(p.transcript_segments) AS s(segment)
    WHERE p.lang_detected IN ('en', 'es', 'de', 'fr', 'ru', 'pt')
),
ranked AS (
    SELECT
        *,
        session_created_at
            + COALESCE(segment_start_seconds, 0) * INTERVAL 1 SECOND
            AS segment_created_at,
        ROW_NUMBER() OVER (
            PARTITION BY lang_detected
            ORDER BY
                session_created_at
                    + COALESCE(segment_start_seconds, 0) * INTERVAL 1 SECOND DESC,
                gamesession_id DESC
        ) AS language_rank
    FROM exploded
    WHERE
        segment_text IS NOT NULL
        AND TRIM(segment_text) <> ''
)
SELECT
    lang_detected,
    gamesession_id,
    user_id,
    game_name,
    model_type,
    session_created_at,
    segment_created_at,
    segment_start_seconds,
    segment_end_seconds,
    lang_probability,
    segment_text
FROM ranked
WHERE language_rank <= 25
ORDER BY
    lang_detected,
    segment_created_at DESC
"""

latest_25_segments = duckdb.sql(query).df()

display(latest_25_segments)
```

```python
segment_counts = (
    latest_25_segments
    .groupby("lang_detected")
    .size()
    .reset_index(name="segment_count")
    .sort_values("segment_count", ascending=False)
)

display(segment_counts)

print("Total segments:", len(latest_25_segments))
```

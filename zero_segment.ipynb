```python
%pip -q install duckdb

from pathlib import Path

import duckdb
import pandas as pd
from google.colab import drive

if not Path("/content/drive/MyDrive").exists():
    drive.mount("/content/drive")

BASE_DIR = Path("/content/drive/MyDrive/Language Detection")

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
pd.set_option("display.expand_frame_repr", False)

con = duckdb.connect()

print("Ready")
```

```python
PARQUET_PATH = BASE_DIR / "sessions_lang_transcript_2026-08-04_2026-08-05.parquet"

if not PARQUET_PATH.is_file():
    raise FileNotFoundError(PARQUET_PATH)

schema = con.execute(
    "DESCRIBE SELECT * FROM read_parquet(?)",
    [PARQUET_PATH.as_posix()],
).df()

result_2026_08_04 = con.execute(
    """
    SELECT
        *,
        len(transcript_segments) AS segment_count
    FROM read_parquet(?)
    WHERE
        transcript_segments IS NULL
        OR len(transcript_segments) = 0
    """,
    [PARQUET_PATH.as_posix()],
).df()

print("FILE:", PARQUET_PATH.name)
print(f"SIZE: {PARQUET_PATH.stat().st_size / 1024**2:.2f} MB")
print(f"ZERO-SEGMENT SESSIONS: {len(result_2026_08_04):,}")

print("\nSCHEMA")
display(schema)

print("\nALL ZERO-SEGMENT ROWS — ALL ORIGINAL COLUMNS")
display(result_2026_08_04)
```
